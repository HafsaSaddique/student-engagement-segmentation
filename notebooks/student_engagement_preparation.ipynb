```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/raw_student_data.csv")

df.head()
```

```python
df.shape
```

```python
df.columns.tolist()
```

```python
df.isnull().sum()
```

```python
df["DaysPresent"] = df["DaysPresent"].fillna(df["DaysPresent"].median())
```

```python
df["AttendanceRate"] = df["DaysPresent"] / df["DaysTotal"]
```

```python
df[["StudentID", "DaysPresent", "DaysTotal", "AttendanceRate"]].head()
```

```python
df["MathScore"] = df["MathScore"].fillna(df["MathScore"].median())
```

```python
df["AvgGrade"] = df[
    ["MathScore", "EnglishScore", "ScienceScore"]
].mean(axis=1)
```

```python
df[
    ["StudentID", "MathScore", "EnglishScore", "ScienceScore", "AvgGrade"]
].head()
```

```python
df["Clubs"] = df["Clubs"].fillna("")
```

```python
df["ExtracurricularCount"] = df["Clubs"].apply(
    lambda x: 0 if x == "" else len(x.split(";"))
)
```

```python
df[
    ["StudentID", "Clubs", "ExtracurricularCount"]
].head(10)
```

```python
cleaned_df = df[
    [
        "StudentID",
        "AttendanceRate",
        "AvgGrade",
        "ExtracurricularCount"
    ]
].copy()
```

```python
cleaned_df.head()
```

```python
cleaned_df.isnull().sum()
```

```python
cleaned_df.shape
```

```python
cleaned_df.to_csv(
    "../data/cleaned/cleaned_student_data.csv",
    index=False
)
```
